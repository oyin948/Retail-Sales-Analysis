```python
#convert order data and ship date to datetime format. format="mixed" allows pandas to correctly handle different date formats. dayfirst=True  tells pandas that the first number represents the day
df["Order Date"] = pd.to_datetime(df["Order Date"], format = "mixed", dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format = "mixed", dayfirst=True)
#check that both data columns have been converted to datetime
df[["Order Date", "Ship Date"]].dtypes
```

```python
#display the records where postal code is missing
missing_postel = df[df["Postal Code"].isnull()]
#display relevant information about the missing postal code records
missing_postel[["Order ID","City","State","Postal Code","Region"]]
```

```python
#calculate the first quartile and third quartile of sales
q1 = df["Sales"].quantile(0.25)
q3 = df["Sales"].quantile(0.75)
#calculate the interquartile range
iqr = q3-q1
#calculate the lower and upper boundaries for potential outliers
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
#display the boundaries
print("Lower Bound: ", lower_bound)
print("Upper Bound: ", upper_bound)
```

```python
#identify potential sales outliers using the iqr method
sales_outliers = df[~df["Sales"].between(lower_bound, upper_bound)]
#count the number of potential sales outliers
sales_outlier_count = sales_outliers.shape[0]
#display the number of potential sales outliers
print("Number of potential sales outliers: ", sales_outlier_count)
```

DATA CLEANING

```python
#create a copy of the dataset for cleaning. This keeps the original DataFrame seperate from the cleaned version 
cleaned_df = df.copy()
```

```python
#ensure order date and ship date are stored as datetime values
cleaned_df["Order Date"] = pd.to_datetime(cleaned_df["Order Date"], dayfirst=True)
cleaned_df["Ship Date"] = pd.to_datetime(cleaned_df["Ship Date"], dayfirst=True)
cleaned_df[["Order Date", "Ship Date"]].dtypes
```

```python
#identity text columns in the cleaned dataset
text_columns = cleaned_df.select_dtypes(include="object").columns
#remove leading and trailing spaces from text values
for column in text_columns:
    cleaned_df[column] = cleaned_df[column].str.strip()
cleaned_df
```

```python
#convert postal code to a text column so it can contain both postal code the label "unknown"
cleaned_df["Postal Code"] = cleaned_df["Postal Code"].astype("Int64").astype("string")
#replace missing code values with "unknown"
cleaned_df["Postal Code"] = cleaned_df["Postal Code"].fillna("unknown")
#check the number of records assigned the "unknown" postal code label 
(cleaned_df["Postal Code"] == "unknown").sum()
```

Sales outlier treatment: The IQR method identified 1,145 potential sales outliers. These observations were retained because statistical outliers are not necessarily data errors, and there was no evidence that these values were incorrect.

```python
#check the final number of rows and columns
print("Dataset shape: ", cleaned_df.shape)
#check for duplicate rows
print("Duplicate rows: ", cleaned_df.duplicated().sum())
#check for missing values
print("Missing values: ", cleaned_df.isnull().sum())
#check the date column date types
print("Date data types: ", cleaned_df[["Order Date","Ship Date"]].dtypes)
```

Data Cleaning Summary

The dataset initially contained 9,800 records and 18 columns. Data quality checks were performed to examine missing values, duplicate records, incorrect data types, categorical values, negative sales values, missing postal codes, and potential numerical outliers. 

The dataset contained 11 missing postal code values. These records were retained because the other information contained in the transactions remained useful. Since the correct postal codes could not be reliably determined from the dataset, the missing values were replaced with "unknown". 

The order date and ship date columns were converted to appropriate datatime formats. Text columns were stripped of unnecessary leading and trailing spaces.

No completely duplicate rows were found and no negative value were identified. Potential sales outliers were detected using the IQR method but retained because there was no evidence that they represented data-entry errors.

After cleaning, the dataset remained at 9,800 rows and 18 columns and was saved as "cleaned_retail_sales.csv".
